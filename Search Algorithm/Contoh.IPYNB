```python

from collections import deque

# Membuat graf
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}

def bfs(graph, start, goal):
    queue = deque([start])
    visited = {start}
    parent = {start: None}

    while queue:
        current = queue.popleft()

        # Jika goal ditemukan
        if current == goal:
            path = []

            while current is not None:
                path.append(current)
                current = parent[current]

            return path[::-1]

        # Mengunjungi tetangga
        for neighbor in graph[current]:
            if neighbor not in visited:
                visited.add(neighbor)
                parent[neighbor] = current
                queue.append(neighbor)

    return None


# Input dari pengguna
print("=== Breadth-First Search (BFS) ===")
print("Simpul yang tersedia:", ", ".join(graph.keys()))

start = input("Masukkan simpul start: ").strip().upper()
goal = input("Masukkan simpul goal: ").strip().upper()

# Validasi input
if start not in graph or goal not in graph:
    print("Error: Simpul start atau goal tidak tersedia.")
else:
    result = bfs(graph, start, goal)

    if result:
        print("Jalur ditemukan:", " -> ".join(result))
        print("Jumlah langkah:", len(result) - 1)
    else:
        print("Jalur tidak ditemukan.")
```
