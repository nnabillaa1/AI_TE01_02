# Rule-Based Medical Advisor

A simple expert system represents domain knowledge as facts and IF-THEN rules. It demonstrates how a knowledge base can produce an explainable recommendation.

## Lab objective

Run the cells in order, inspect the result, and modify the facts, rules, or parameters.

```python
facts = {"fever", "cough", "fatigue"}
rules = [
    ({"fever", "cough"}, "possible_flu"),
    ({"possible_flu", "fatigue"}, "recommend_rest"),
    ({"possible_flu"}, "consult_doctor"),
]

fired = True
while fired:
    fired = False
    for conditions, conclusion in rules:
        if conditions <= facts and conclusion not in facts:
            facts.add(conclusion)
            fired = True
            print(f"Rule fired: {sorted(conditions)} -> {conclusion}")

print("Derived knowledge:", sorted(facts))
```
